```python
# positional argument 
def student(name,age,course):
    print("name",name)
    print("age",age)
    print("course",course)
student(24,"sri","ai")
```

```python
# DEFAULT ARGUMENT
def data(name="guest"):
    print(f"welcome {name}")
data()
data("priya")
```

```python
#keyword arguments
def details(name,city,age,interest):
    print(name)
    print(city)
    print(age)
    print(interest)
details(name="vishnu",age=27,city="cbe",interest="cricket")
```

```python
#variable length positional arguments
def total(*args):
    return sum(args)
total(90,89,78,67,56,56,45,44,34,12,34,56,7,89,9)
```

```python
#variable length keyword arguments 
def data (**kwargs):
    for key,value in kwargs.item
    
data(name="priya",age=24,city="cbe",country="india",interest="dancing",hobby="cooking")
```
