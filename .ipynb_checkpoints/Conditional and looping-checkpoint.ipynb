#if else
if condition:
    block of code
else:
    block of code

```python
x=10

if x<10:
    print("x is less than 10")
else:
    print("x is 10 or greater than 10")
```

```python
number1=int(input("Enter the number1"))

if number1%2==0:
    # print(number1,"is even")
    print(f"{number1} is even")
else:
    print(f"{number1} is odd")
```

```python
number2=int(input("Enter the number"))

if number2<0:
    print("its negative")
elif number2==0:
    print("its zero")
else:
    print("its positive")
```

```python
marks=int(input("Enter the marks (0-100)"))

if marks>=90 and marks<=100:
    print("Grade A")
elif marks>=80 and marks<=89:
    print("Grade B")
elif marks>=70 and marks<=79:
    print("Grade C")
elif marks>=60 and marks<=69:
    print("Grade D")
elif marks<60:
    print("Grade F")
else:
    print("Invalid marks entered")
```

```python
number=int(input("Enter the number"))

if number%3==0 and number%5==0:
    print(f"{number} divided by both 5 and 3")
else:
    print("its not")
```

```python
marks=int(input("Enter your 12th marks"))

if marks>500:
    print("You're eligible to attend the interview")
    first_round_score=int(input("Enter the first round score"))
    if first_round_score>30:
        print("you've cleared the first round")
        hr=10
        if hr>5:
            print("Congrats, you've got the offer")
        else:
            print("get back to you")
    else:
        print("Sorry,you're not cleared")
else:
    print("sorry, you're not eligible for developer role")
    if marks>=400:
        print("You can attend the role:tester")
    else:
        print("Next time you can try")
        
```

```python
user_input=int(input("Enter the number"))

if user_input>0:
    if user_input%2==0:
        print(f"{user_input} is even")
    else:
        print(f"{user_input} is odd")
elif user_input<0:
    if abs(user_input)%5==0:
        print(f"{user_input} absolute value is multiples of 5")
    else:
        print("its not multiples of 5")

else:
    print("Zero is neither positive nor negative.")
```

```python
user_name=input("Enter the username")
password=input("Enter the password")

if user_name=="admin" and password=="admin123":
    print("Login successful")
elif user_name=="admin" and password!="admin123":
    print("Wrong password")
else:
    print("User not found")
```
